```python
# 1. Definición de origen y carga (Boilerplate de clase)
url = 'https://raw.githubusercontent.com/selva86/datasets/master/Carseats.csv'
df = pd.read_csv(url)

# 2. Configuración del lienzo y creación del gráfico
plt.figure(figsize=(10, 6))
sb.barplot(data=df, x='ShelveLoc', y='Sales', palette='viridis', hue='ShelveLoc')

# 3. Personalización (Capa Matplotlib)
plt.title('Ventas Promedio por Calidad de Ubicación', fontweight="bold")
plt.xlabel('Ubicación en Estante')
plt.ylabel('Ventas (miles de unidades)')
plt.xticks(rotation=0) # Etiquetas horizontales
plt.tight_layout()

# 4. Renderizado final
plt.show()
```

```python
# 1. Matriz de Ventas:
ventas = np.array([
    [2, 3, 2],   # Producto A101 Harina
    [4, 1, 7],   # Producto B310 Gaseosa
    [5, 0, 4]    # Producto L150 Detergente
])
productos = ["A101 Harina", "B310 Gaseosa", "L150 Detergente"]
sucursales = ["Neuquén", "Bariloche", "Trelew"]

# 2. Operaciones con axis
total_por_sucursal = ventas.sum(axis=0)  # axis=0: Suma vertical ⬇️ (colapsa las filas)
total_por_producto = ventas.sum(axis=1)  # axis=1: Suma horizontal ➡️ (colapsa las columnas)

# 3. Armamos la tabla de presentación (DataFrame)
tablero = pd.DataFrame(ventas, index=productos, columns=sucursales)

# Agregamos la columna de TOTAL por producto (axis=1)
tablero["TOTAL Producto (axis=1)"] = total_por_producto

# Agregamos la fila de TOTAL por sucursal (axis=0) y el Total General al final
tablero.loc["TOTAL Sucursal (axis=0)"] = list(total_por_sucursal) + [ventas.sum()]

print(tablero)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pulp
# 1. Creamos el problema indicando si Maximizamos o Minimizamos
prob = pulp.LpProblem("Optimizacion_Fabrica", pulp.LpMaximize)

# 2. Definimos las variables de decisión (con límite inferior 0)
x1 = pulp.LpVariable('Sillas', lowBound=0)
x2 = pulp.LpVariable('Mesas', lowBound=0)

# 3. Agregamos la Función Objetivo directamente
prob += 50 * x1 + 80 * x2, "Ganancia_Total"

# 4. Agregamos las restricciones
prob += 2 * x1 + 1 * x2 <= 100, "Carpinteria"
prob += 1 * x1 + 1 * x2 <= 80,  "Pintura"

# 5. Resolvemos
prob.solve()

# Muestra el resultado de las variables y el óptimo
print("Estado:", pulp.LpStatus[prob.status])
print("Sillas a fabricar:", x1.varValue)
print("Mesas a fabricar :", x2.varValue)
print("Ganancia Máxima  :", pulp.value(prob.objective))
```
